```python
import matplotlib.pyplot as plt
```

```python
drinks = ['Tea','Hot coffee', 'Cold coffee','Juice']
summer = [70,50,80,60]
winter = [90,70,30,50]
```

#### Bar plot
1. Comparing caregories
2. Showing frequency or count.

```python
plt.figure(figsize=(4,3))
plt.bar(drinks,summer,color ='orange')
plt.title('Drinks sales')
plt.xlabel('Drinks')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(4,3))
plt.bar(drinks,winter,color ='blue')
plt.title('Drinks sales')
plt.xlabel('Drinks')
plt.ylabel('Count')
plt.show()
```
