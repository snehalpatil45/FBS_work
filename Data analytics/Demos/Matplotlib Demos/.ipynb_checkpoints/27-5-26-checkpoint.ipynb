### Histogram
1. Distribution of numerical data
2. Frequency of value ranges
3. Normal distribution
4. Skewness
5. Outliers

```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("C:/Users/ASUS/Desktop/FBS/Data analytics/Datasets/Mall_Customers.csv")
data.head()
```

```python
male = data[data['Gender'] == 'Male']['Age']
male
```

```python
female = data[data['Gender'] == 'Female']['Age']
female
```

```python
import matplotlib.pyplot as plt
from matplotlib import style
style.use('ggplot')
plt.style.use('dark_background')
```

```python
plt.hist([male,female],rwidth = 0.8,label =['Male','Female'],color =['skyblue','m'])
plt.legend()
plt.show()
```

#### Line plot
1. Showing the trends
2. Time series analysis

```python
overs = np.arange(1,21)
team1 = np.random.randint(0,37,20)
team2 = np.random.randint(0,37,20)
team1,team2
```

```python
team1_score = np.cumsum(team1)
team2_score = np.cumsum(team2)
team1_score,team2_score
```

```python
plt.figure(figsize =(6,3))
plt.plot(overs,team1_score,color='r',marker='o',label ='Team 1')
plt.plot(overs,team2_score,color='y',marker='*',label ='Team 2')
plt.grid(alpha=0.3,linestyle='--')
plt.xticks(overs)
plt.legend()
plt.title('IPL Match Score Comparison')
plt.xlabel('Overs')
plt.ylabel('Runs')
plt.tight_layout()
plt.savefig('IPL Match Score Comparison.jpg')
plt.show()
```

#### Scatter plot
1. Relationship between two variables
2. correlation analysis

```python
plt.figure(figsize=(6,3))
plt.scatter(overs,team1,label='Team 1',color='r',marker='>',edgecolor='y',s=100)
plt.scatter(overs,team2,label='Team 2',color='b',marker='*',edgecolor='m',s=100)
plt.grid(alpha=0.3,linestyle='--')
plt.xticks(overs)
plt.legend()
plt.show()
```

#### Pie chart
1. Showing percentage distribution
2. Part to whole comparison

```python
classes =['python','Java','ML','DS','Testing']
class1 = [45,15,35,24,30]
colors =['c','b','r','y','g']
explode = [0.03,0,0.1,0,0]
plt.pie(class1,labels = classes,colors = colors,explode = explode)
plt.show()
```
