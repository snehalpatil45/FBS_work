### Basic Shortcuts of Jupyter Notebook

1. **A** → Add cell above  
2. **B** → Add cell below  
3. **DD** → Delete selected cell  
4. **M** → Change cell to Markdown  
5. **Shift + Enter** → Execute cell and add new cell  
6. **Ctrl + Enter** → Execute current cell only  
7. **Y** → Change cell to Code  
8. **X** → Cut cell  
9. **C** → Copy cell  
10. **V** → Paste cell below  
11. **Z** → Undo deleted cell  
12. **Esc** → Command mode  
13. **Enter** → Edit mode
