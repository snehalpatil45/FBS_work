```python
import pandas as pd
import numpy as np
```

```python
students = [f"Student_{i}" for i in range(1, 31)]

subjects = ["Physics", "Chemistry", "Biology","Mathematics","Geography", "English"]

marks = np.random.randint(30, 100,(30, 6))

df = pd.DataFrame(marks, columns=subjects)

df.insert(0, "Student_Name", students)

df["Total"] = df[subjects].sum(axis=1)

df["Percentage"] = (df["Total"] / 600) * 100

print(df)
```

```python
df.to_excel("student_marks.xlsx", index=False)
```
