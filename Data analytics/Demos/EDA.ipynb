```python
q1 = iris['Sepal_Width'].quantile(0.25)
q1
```

```python
q3  = iris['Sepal_Width'].quantile(0.75)
q3
```

```python
iqr = q3 - q1
iqr
```

```python
upper_bound = q3 + 1.5* iqr
lower_bound = q1 - 1.5* iqr
```

```python
iris[iris['Sepal_Width'] > upper_bound]
```

```python
iris2 = iris[(iris['Sepal_Width']>= lower_bound) & (iris['Sepal_Width'] <= upper_bound)]
```

```python
sns.boxplot(data = iris2)
```
