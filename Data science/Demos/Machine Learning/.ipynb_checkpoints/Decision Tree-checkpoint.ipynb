#### decision tree used for both classification and regression

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("C:/Users/ASUS/Desktop/FBS/Data analytics/Datasets/Iris.csv")
data.head()
```

```python
X = data.iloc[:,:-1]
X.head()
```

```python
X.info()
```

```python
y = data['Species']
y.head()
```

```python
y.info()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size = 0.2,random_state = 10)
```

```python
from sklearn.tree import DecisionTreeClassifier
dtree = DecisionTreeClassifier()
```

```python
dtree.fit(X_train,y_train)
```

```python
train_predict = dtree.predict(X_train)
```

```python
test_predict = dtree.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y_train,train_predict)
```

```python
accuracy_score(y_test,test_predict)
```

```python
from sklearn.tree import plot_tree
plt.figure(figsize = (7,7))
plot_tree(dtree,class_names = data['Species'].unique(),feature_names = X.columns,filled = True)
plt.show()
```
