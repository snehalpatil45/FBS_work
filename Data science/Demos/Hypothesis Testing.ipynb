```python
import numpy as np
from scipy  import stats
```

```python
data = [70,65,67,82,80,64,83]
```

```python
#H0 - Null hypothesis
mu = 70
```

```python
# perform test
# t-test = sample_size < 30
# z-test = sample_size >= 30
# chi-test = sample is categorical
# statistics = how far is my sample result from what i expected(H0)
```

```python
statistics,p_value = stats.ttest_1samp(data,mu)
```

```python
print(p_value)
```

```python
alpha = 0.05
```

```python
if p_value < alpha:
    print('Reject the null hypothesis')
else:
    print('Failed to reject the null hypothesis')
```
