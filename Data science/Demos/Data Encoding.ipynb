#### Ordinal Data

##### ordinal encoding

```python
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories = [['Low','Medium','High']])
```

```python
data = np.array([['Low'],['Medium'],['High']])
oe.fit_transform(data)
```
