### steps
1. Tokenization
2. Identiy Part of speech
3. Stemming and lemmitization
4. Stop words removal
5. Entity recognization

```python
import spacy
```

```python
nlp = spacy.load('en_core_web_sm')
```

```python
text ='''Dr. Smith lives in the U.S. and works at New York University. 
He specializes in A.I. and has published over 3.14 papers every year. 
He often travels to Washington D.C. for conferences, workshops, etc. 
Recently, he received an award from the A.I. Society.'''
```

```python
doc = nlp(text)
doc
```

```python
# tokenization and pos
for sen in doc.sents:
    print(sen)
```

```python
for token in doc:
    print(token,' ',token.pos_)  # tokeniztion and pos
```

```python
# stemming 
import nltk
from nltk.stem import PorterStemmer
obj = PorterStemmer()
```

```python
text = 'I studied Data Science'
```

```python
li = text.split(' ')
```

```python
obj.stem('meeting')
```

```python
for token in li:
    print(obj.stem(token))
```

```python
doc2 = nlp(text)
```

```python
#lemmitization
for token in doc2:
    print(token,':',token.lemma_)
```

```python
# Entity recognization
for token in doc.ents:# entities
    print(token ,':',token.label_,':',spacy.explain(token.label_))
```

```python
# stop words removal
stop_words = nlp.Defaults.stop_words
#len(nlp.Defaults.stop_words)
```

```python
text = 'We are learning NLP.Today is not holiday.'
```

```python
import re
```

```python
res = re.findall(r'\w+',text)
text = ' '.join(res)
text
```

```python
result = []
doc = nlp(text)
for token in doc:
    token = str(token)
    token = token.lower()
    if token not in stop_words:
        result.append(token)
result
```
