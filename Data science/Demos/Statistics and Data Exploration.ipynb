```python
q1 = df['sold_qty'].quantile(0.25) # for 25 percentile
q1
```

```python
q3 = df['sold_qty'].quantile(0.75) # for 75 percentile
q3
```

```python
iqr = q3 - q1
iqr
```

```python
lower_bound = q1 - (1.5 * iqr)
lower_bound
```

```python
upper_bound = q3 + (1.5 * iqr)
upper_bound
```
